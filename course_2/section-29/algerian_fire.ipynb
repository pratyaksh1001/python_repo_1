```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("Algerian_forest_fires_dataset.csv")
```

```python
df
```

```python
df = df[df["Classes  "] != "Classes  "]
```

```python
df=df.dropna()
```

```python
df.columns=df.columns.str.strip()
```

```python
df["Classes"]=df["Classes"].str.strip()
```

```python
df["Classes"].value_counts()
```

```python
df["Classes"].value_counts().plot(kind="pie")
```

```python
df.dtypes
```

```python
columns=df.columns
```

```python
for i in columns:
    if i!="Classes":
        df[i]=df[i].astype(float)
```

```python
df.dtypes
```

```python
df
```

```python
df.replace({"fire":1,"not fire":0},inplace=True)
```

```python
df=df.drop("year",axis=1)
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
x=df.drop("Classes",axis=1)
y=df["Classes"]
```

```python
scaler=StandardScaler()
scaler.fit(x)
scaler.transform(x)
```

```python
linear_model=LinearRegression()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
linear_model.fit(x_train,y_train)
```

```python
linear_model.score(x_test,y_test)
```

```python
from sklearn.linear_model import ElasticNetCV
```

```python
elastic_model=ElasticNetCV(cv=5)
```

```python
elastic_model.fit(x_train,y_train)
```

```python
elastic_model.score(x_test,y_test)
```

```python
y_preds=elastic_model.predict(x_test)
```

```python
y_preds
```

```python
y_preds=np.round(y_preds)
```

```python
res=y_preds==y_test
```

Accuracy of the Elasticnet cross validated model

```python
np.count_nonzero(res)/len(res)*100
```
